```python
from evaluate import load
import pandas as pd
import nltk
import json
```

```python
CACHE_DIR = "cache"

nltk.data.path.append(CACHE_DIR)
rouge = load('rouge', cache_dir=CACHE_DIR)
bleu = load('bleu', cache_dir=CACHE_DIR)
meteor = load('meteor', cache_dir=CACHE_DIR)
bertscore = load("bertscore", cache_dir=CACHE_DIR)
```

```python
def calculate_scores(true_df, pred_df, true_col='Decision', pred_col='Decision'):
    
    results = {}
    results['rouge'] = rouge.compute(predictions=pred_df[pred_col],references= true_df[true_col])
    print('Rouge Done')
    results['bleu'] = bleu.compute(predictions=pred_df[pred_col],references= true_df[true_col])
    print('Bleu Done')
    results['meteor'] = meteor.compute(predictions=pred_df[pred_col],references= true_df[true_col])
    print('Meteor Done')
    results['bertscore'] = bertscore.compute(predictions=pred_df[pred_col],references= true_df[true_col], lang='en', batch_size = 64)
    print('BertScore Done')
    cols = ['precision', 'recall', 'f1']
    for c in cols:
        results['bertscore'][c] = pd.Series(results['bertscore'][c]).mean()
    
    results['AverageTime'] = pred_df['Time'].mean()
    results['AverageGeneratedTokens'] = pred_df['GeneratedTokens'].mean()
    return results
```

### Prompting gemma-3-4b-it CDtest

```python
# Load true and predicted data
true_file = "DRAFT/Retrieval/CDtest.jsonl"
pred_file = "DRAFT/Prompting/Results/gemma-3-4b-it-CDtest-results.jsonl"

true_df = pd.read_json(true_file, lines=True)
true_df = true_df["Anchor"].apply(pd.Series)
pred_df = pd.read_json(pred_file, lines=True)

# For testing purposes, limit to first 10 rows. Comment out in real experiments.
# true_df = true_df.head(10)
# pred_df = pred_df.head(10)

results = calculate_scores(true_df, pred_df)
```

```python
# Save results to appropriate JSON file
approach = "Prompting"
model = "gemma-3-4b-it"
dataset = "CDtest"
output_path = f"DRAFT/Results/{approach}_{model}_{dataset}.json"
with open(output_path, 'w') as f:
    json.dump(results, f, indent=2)
```
